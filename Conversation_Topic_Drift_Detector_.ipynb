```python

# Conversation Topic Drift Detector

from collections import Counter
import re

topic_keywords = {
    "Education": [
        "student", "exam", "study", "college",
        "teacher", "assignment"
    ],
    "Technology": [
        "python", "software", "computer", "coding",
        "program", "technology"
    ],
    "Health": [
        "doctor", "health", "medicine", "hospital",
        "exercise", "patient"
    ],
    "Travel": [
        "travel", "train", "flight", "hotel",
        "trip", "ticket"
    ]
}

def detect_topic(text):
    words = re.findall(r"\b[a-zA-Z]+\b", text.lower())
    scores = {}

    for topic, keywords in topic_keywords.items():
        scores[topic] = sum(
            1 for word in words if word in keywords
        )

    best_topic = max(scores, key=scores.get)

    if scores[best_topic] == 0:
        return "Unknown"

    return best_topic


def analyze_conversation(messages):
    print("=" * 45)
    print("CONVERSATION TOPIC DRIFT DETECTOR")
    print("=" * 45)

    previous_topic = None
    drift_count = 0

    for i, message in enumerate(messages, start=1):
        current_topic = detect_topic(message)

        print(f"\nMessage {i}: {message}")
        print(f"Detected topic: {current_topic}")

        if previous_topic is not None:
            if current_topic != previous_topic:
                print(
                    f"Topic shift: {previous_topic} -> {current_topic}"
                )
                drift_count += 1

        previous_topic = current_topic

    print("\nTotal topic shifts:", drift_count)
    print("=" * 45)


# Sample conversation
messages = [
    "Students are preparing for the college exam.",
    "The teacher gave a new study assignment.",
    "I am learning Python programming today.",
    "Software coding is interesting.",
    "I need to visit the doctor for my health."
]

analyze_conversation(messages)
```
